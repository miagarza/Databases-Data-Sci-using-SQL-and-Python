```python
%sql select name_of_school, AVERAGE_STUDENT_ATTENDANCE from 'ChicagoPublicSchools.csv' \
order by AVERAGE_STUDENT_ATTENDANCE desc nulls last limit 10;
```

```python
%sql select name_of_school, AVERAGE_STUDENT_ATTENDANCE from 'ChicagoPublicSchools.csv' \
order by AVERAGE_STUDENT_ATTENDANCE limit 5;
```

```python
%sql select name_of_school, replace(AVERAGE_STUDENT_ATTENDANCE, '%','') from 'ChicagoPublicSchools.csv' \
order by AVERAGE_STUDENT_ATTENDANCE limit 5;
```

```python
%sql select name_of_school, AVERAGE_STUDENT_ATTENDANCE from 'ChicagoPublicSchools.csv' where AVERAGE_STUDENT_ATTENDANCE<70;
```

```python
%sql select COMMUNITY_AREA_NAME, sum(College_Enrollment) FROM 'ChicagoPublicSchools.csv'\
GROUP BY COMMUNITY_AREA_NAME ;
```

```python
%sql select COMMUNITY_AREA_NAME, sum(College_Enrollment) FROM 'ChicagoPublicSchools.csv'\
GROUP BY COMMUNITY_AREA_NAME ORDER BY College_Enrollment asc LIMIT 5;
```

```python
%sql select NAME_OF_SCHOOL, SAFETY_SCORE from 'ChicagoPublicSchools.csv'\
WHERE SAFETY_SCORE!='NONE' ORDER BY SAFETY_SCORE LIMIT 5;
```
